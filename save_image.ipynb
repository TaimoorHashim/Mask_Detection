```python
import cv2
import os

# Define classes
classes = ["with_mask", "without_mask"]

# Create dataset folders
for cls in classes:
    os.makedirs(f"dataset/{cls}", exist_ok=True)

def open_camera():
    # DirectShow prevents Windows MSMF error -1072875772
    for index in [0, 1, 2]:
        cap = cv2.VideoCapture(index, cv2.CAP_DSHOW)
        if cap.isOpened():
            ret, frame = cap.read()
            if ret and frame is not None:
                print(f"Connected to camera index {index} via DirectShow.")
                return cap
            cap.release()

    for index in [0, 1]:
        cap = cv2.VideoCapture(index)
        if cap.isOpened():
            ret, frame = cap.read()
            if ret and frame is not None:
                print(f"Connected to camera index {index} via default backend.")
                return cap
            cap.release()
    return None

cap = open_camera()

if cap is None:
    print("Could not access webcam. Ensure no other application (Zoom, Teams, Camera) is using it.")
else:
    # Choose class to capture: "with_mask" or "without_mask"
    class_name = "without_mask"

    count = len(os.listdir(f"dataset/{class_name}"))
    print(f"Collecting images for '{class_name}'. Press 's' to save, 'q' to quit.")

    while True:
        ret, frame = cap.read()
        if not ret or frame is None:
            print("Failed to grab frame.")
            break

        cv2.imshow("Capture - Press 's' to Save, 'q' to Quit", frame)
        key = cv2.waitKey(1) & 0xFF

        if key == ord('s'):
            path = f"dataset/{class_name}/{class_name}_{count}.jpg"
            cv2.imwrite(path, frame)
            print(f"Saved: {path}")
            count += 1
        elif key == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()
```
